```python
# Final robust text cleaning function
import ftfy

def fix_encoding(text):
    try:
        return ftfy.fix_text(text)
    except:
        return text

# Apply to both title and description
df["Title"] = df["Title"].apply(fix_encoding)
df["Description"] = df["Description"].apply(fix_encoding)

# Preview
df[["Title", "Description"]].head(10)
```

```python
from openai import OpenAI
import os
import json
from dotenv import load_dotenv

# Load your API key from .env
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")

# Initialize OpenAI client
client = OpenAI(api_key=api_key)

# Define the function
def extract_with_gpt(article):
    print("🔍 Sending prompt to GPT for:", article["Title"][:60])

    prompt = f"""
You are an AI assistant that extracts structured business insights from news articles.

From the following article, extract the following information as valid JSON.

If any field is not explicitly stated, **infer it** using your general knowledge about the company. Never leave fields empty. Always respond in JSON.

Format:

{{
    "company": "main company name mentioned",
    "ceo": "CEO or founder (use your knowledge if not stated)",
    "category": "choose best matching from: AI, Education, Finance, Healthcare, IT, Others",
    "key_insight": [
        "bullet point 1 - key event or issue",
        "bullet point 2 - reaction, action or impact",
        "bullet point 3 - strategic or market implication"
    ],
    "investment_location": "country or region (infer if needed)",
    "sentiment": "Positive / Negative / Neutral",
    "relevance_score": "Rate from 1 to 5 stars"
}}

Title: {article['Title']}
Description: {article['Description']}
"""

    try:
        response = client.chat.completions.create(
            model="gpt-3.5-turbo",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.3
        )

        content = response.choices[0].message.content.strip()

        try:
            return json.loads(content)
        except json.JSONDecodeError:
            return {"raw_output": content, "error": "JSON parsing failed"}

    except Exception as e:
        return {"error": str(e)}
```

```python
import time
import pandas as pd

# List to collect all outputs
extracted_data = []

# Loop through each article
for idx, row in df.iterrows():
    print(f"\n🔄 Processing article {idx + 1}/{len(df)}")

    # Run GPT extractor
    result = extract_with_gpt(row)

    # Flatten and organize fields for CSV
    output = {
        "title": row.get("Title", ""),
        "description": row.get("Description", ""),
        "link": row.get("Link", ""),
        "company": result.get("company"),
        "ceo": result.get("ceo"),
        "category": result.get("category"),
        "investment_location": result.get("investment_location"),
        "sentiment": result.get("sentiment"),
        "relevance_score": result.get("relevance_score"),
    }

    # Handle 3-part insight extraction
    insights = result.get("key_insight", [])
    for i in range(3):
        output[f"insight_{i+1}"] = insights[i] if i < len(insights) else None

    extracted_data.append(output)

    # Optional: sleep to avoid rate limits
    time.sleep(1.1)

# Convert to DataFrame
final_df = pd.DataFrame(extracted_data)

# Save to CSV
final_df.to_csv("gpt_insights_latam.csv", index=False)
print("✅ All insights saved to gpt_insights_latam.csv")
```

```python
# Load the CSV file
df = pd.read_csv("gpt_insights_latam.csv")

# Combine insights
df["insights"] = df[["insight_1", "insight_2", "insight_3"]].apply(
    lambda x: "\n- " + "\n- ".join(filter(pd.notna, x)), axis=1
)

# Remove "stars" from relevance_score
df["relevance_score"] = df["relevance_score"].str.replace("stars", "", case=False).str.strip()

# Drop old insight columns
df = df.drop(columns=["insight_1", "insight_2", "insight_3"])

# Save the final cleaned CSV
df.to_csv("gpt_insights_latam_cleaned.csv", index=False)
print("✅ Cleaned CSV saved as gpt_insights_latam_cleaned.csv")
```
